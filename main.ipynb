```python

import pandas as pd
from sklearn.datasets import load_iris
from scipy import stats

# Carrega o dataset Iris (UCI ML Repository, ID 53)
iris = load_iris(as_frame=True)
df = iris.frame.copy()
df['species'] = df['target'].map(dict(enumerate(iris.target_names)))

print("Primeiras linhas do dataset:")
print(df[['sepal length (cm)', 'petal length (cm)', 'species']].head())
print("\nInfo geral:")
print(df.dtypes)
print("\nTotal de registros:", len(df))

# ---- Variável 1: petal length (cm) -> QUANTITATIVA CONTÍNUA ----
petal = df['petal length (cm)']
print("\n=== Variável: petal length (cm) — Quantitativa Contínua ===")
print("Média:", round(petal.mean(), 3))
print("Mediana:", round(petal.median(), 3))
moda_petal = petal.mode()
print("Moda:", moda_petal.tolist(), "(freq:", (petal == moda_petal[0]).sum(), ")")
print("Desvio padrão:", round(petal.std(), 3))
print("Mínimo/Máximo:", petal.min(), "/", petal.max())

# ---- Variável 2: species -> QUALITATIVA NOMINAL ----
species = df['species']
print("\n=== Variável: species — Qualitativa Nominal ===")
print("Contagem de frequência:")
print(species.value_counts())
moda_species = species.mode()
print("Moda:", moda_species.tolist())
print("(Média e mediana não se aplicam a variáveis qualitativas nominais)")
```
